## **Notebook de Ingestão de Dados para a Camada Bronze**

**Este é um notebook que criará as tabelas na camada bronze da arquitetura.**

Os dados serão carregados de maneira crua (raw data), sem nenhuma transformação neste ponto.

**Importando as bibliotecas:**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

%md
**Criando os DataFrames a partir dos arquivos:**

- Viagens (arquivos Parquet):
    - janeiro
    - fevereiro
    - março
- Zonas (arquivo CSV)

```python
path = [
    "/Volumes/nyc_taxi_data/raw/landing/trips/yellow_tripdata_2025-01.parquet",
    "/Volumes/nyc_taxi_data/raw/landing/trips/yellow_tripdata_2025-02.parquet",
    "/Volumes/nyc_taxi_data/raw/landing/trips/yellow_tripdata_2025-03.parquet"
]


#lendo o dataframe a partir do path
df_viagens = spark.read.parquet(*path)

df_zonas = spark.read.csv("/Volumes/nyc_taxi_data/raw/landing/lookup/taxi_zone_lookup.csv", inferSchema=True, header=True)
```

**Criando as tabelas:**

```python
#Tabela de Viagens
df_viagens.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("nyc_taxi_data.bronze.viagens")


#Tabela de Zonas da Cidade
df_zonas.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("nyc_taxi_data.bronze.zonas")
```
